```python
%sql
CREATE OR REPLACE TABLE dev.sales.delta_compare_orders
USING DELTA
AS
SELECT
    CAST(id AS INT) AS order_id,
    CAST(id % 100 AS INT) AS customer_id,
    date_add(DATE '2026-01-01', CAST(id % 10 AS INT)) AS order_date,
    CAST(id % 500 AS INT) AS amount,
    CASE 
        WHEN id % 3 = 0 THEN 'PAID'
        WHEN id % 3 = 1 THEN 'PENDING'
        ELSE 'FAILED'
    END AS status,
    CASE
        WHEN id % 4 = 0 THEN 'PL'
        WHEN id % 4 = 1 THEN 'DE'
        WHEN id % 4 = 2 THEN 'UK'
        ELSE 'FR'
    END AS country
FROM range(1000);
```

```python
%sql
CREATE OR REPLACE TABLE dev.sales.iceberg_compare_orders
USING ICEBERG
AS
SELECT
    CAST(id AS INT) AS order_id,
    CAST(id % 100 AS INT) AS customer_id,
    date_add(DATE '2026-01-01', CAST(id % 10 AS INT)) AS order_date,
    CAST(id % 500 AS INT) AS amount,
    CASE 
        WHEN id % 3 = 0 THEN 'PAID'
        WHEN id % 3 = 1 THEN 'PENDING'
        ELSE 'FAILED'
    END AS status,
    CASE
        WHEN id % 4 = 0 THEN 'PL'
        WHEN id % 4 = 1 THEN 'DE'
        WHEN id % 4 = 2 THEN 'UK'
        ELSE 'FR'
    END AS country
FROM range(1000);
```
