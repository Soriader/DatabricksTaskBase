```python
%sql
CREATE OR REPLACE VIEW dev.sales.lineage_paid_orders_v AS
SELECT *
FROM dev.sales.lineage_orders_raw
WHERE status = 'PAID';

SELECT * 
FROM dev.sales.lineage_paid_orders_v
```

```python
%sql
CREATE OR REPLACE VIEW dev.sales.lineage_daily_sales_v AS
SELECT
    order_date,
    country,
    COUNT(*) AS orders_count,
    SUM(amount) AS total_amount
FROM dev.sales.lineage_paid_orders_v
GROUP BY order_date, country;
```

```python
%sql
SELECT *
FROM dev.sales.lineage_daily_sales_v
ORDER BY order_date, country;
```
