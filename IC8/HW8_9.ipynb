```python
%sql

CREATE OR REPLACE TABLE dev.sales.liquid_clustering_drill
USING DELTA
CLUSTER BY AUTO
AS
SELECT /*+ REPARTITION(40) */
    CAST(id AS BIGINT) AS order_id,
    CAST(id % 1000 AS INT) AS customer_id,
    date_add(DATE '2026-01-01', CAST(id % 30 AS INT)) AS order_date,
    CAST(id % 100 AS INT) AS amount,
    CASE 
        WHEN id % 3 = 0 THEN 'PAID'
        WHEN id % 3 = 1 THEN 'PENDING'
        ELSE 'FAILED'
    END AS status,
    CASE
        WHEN id % 4 = 0 THEN 'PL'
        WHEN id % 4 = 1 THEN 'DE'
        WHEN id % 4 = 2 THEN 'UK'
        ELSE 'FR'
    END AS country
FROM range(200000);
```

```python
%sql
SELECT *
FROM dev.sales.liquid_clustering_drill
LIMIT 10;
```
